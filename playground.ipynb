```python
import torch
import torch.nn as nn

# Simple model with Dropout
class MyNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(5, 5)
        self.drop = nn.Dropout(p=0.5)  # 50% chance to zero out

    def forward(self, x):
        x = self.fc(x)
        x = self.drop(x)  # dropout applied here
        return x

# Create model and input
model = MyNet()
x = torch.ones(1, 5)   # input vector of ones

# Case 1: Training mode (default)
model.train()
out_train1 = model(x)
out_train2 = model(x)   # run again
print("Training mode outputs:")
print(out_train1)
print(out_train2)

# Case 2: Evaluation mode
model.eval()
out_eval1 = model(x)
out_eval2 = model(x)   # run again
print("\nEvaluation mode outputs:")
print(out_eval1)
print(out_eval2)
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Sample data
X = [[25, 1], [40, 0], [35, 1], [20, 1]]
y = ["Buy", "Don’t Buy", "Buy", "Buy"]  # Labels

# Train decision tree
clf = DecisionTreeClassifier(max_depth=2, random_state=42)
clf.fit(X, y)

# Visualize
plt.figure(figsize=(8,6))
plot_tree(clf, feature_names=["Age", "MorningCommute"], class_names=clf.classes_, filled=True)
plt.show()
```
